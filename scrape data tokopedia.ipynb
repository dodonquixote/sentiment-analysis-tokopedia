```python
from google_play_scraper import reviews, Sort
import pandas as pd
import time
```

# Scraping Data

```python
app_id = 'com.tokopedia.tkpd'

reviews, continuation_token = reviews(
    app_id,
    lang='id',          # Bahasa Indonesia
    country='id',       # Negara Indonesia
    sort=Sort.NEWEST,   # Urutkan dari terbaru
    count=30000,          # Jumlah review
    filter_score_with=None  # Ambil semua rating
)
```

```python
df = pd.DataFrame(reviews)
print("Jumlah total data:", len(df))
df.head()
```

```python
df = df[['content', 'score']]
print("Jumlah total data:", len(df))
df.head()
```

```python
df.to_csv('tokopedia_reviews.csv', index=False, encoding='utf-8-sig')
print("Selesai! File disimpan sebagai `tokopedia_reviews.csv`")
```
